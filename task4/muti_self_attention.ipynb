```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
B, seq_len, d_model = 1,6,16
h = 2
X = torch.randn(B, seq_len, d_model)
Wq = torch.randn(d_model, d_model)
Wk = torch.randn(d_model, d_model)
Wv = torch.randn(d_model, d_model)
Wo = torch.randn(d_model, d_model)

Q = X @ Wq
K = X @ Wk
V = X @ Wv
```

```python
print(Q.shape,K.shape,V.shape)
```

```python
Q = Q.reshape(B, seq_len, h, d_model//h).transpose(1,2) # [B, h, seq_len, d_k]
K = K.reshape(B, seq_len, h, d_model//h).transpose(1,2)
V = V.reshape(B, seq_len, h, d_model//h).transpose(1,2)
```

```python
print(Q.shape,K.shape,V.shape)
```

```python
attn_score = Q @ K.transpose(-1,-2) / torch.sqrt(torch.tensor(d_model//h))
attn_weights = torch.softmax(attn_score, dim=-1)
out_head = attn_weights @ V
```

```python
out_head = out_head.transpose(1,2).reshape(B, seq_len, d_model)
output = out_head @ Wo
```

```python
print("output shape:", output.shape)
print("attn_weights shape [B, h, seq_q, seq_k]:", attn_weights.shape)
```
